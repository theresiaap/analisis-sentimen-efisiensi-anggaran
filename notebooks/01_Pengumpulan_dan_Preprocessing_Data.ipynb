# ***Text Preprocessing***

# *Data Cleaning*

```python
import re

# Fungsi Data Cleaning
def data_cleaning(text):
    # Menghapus URL
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)

    # Menghapus mention
    text = re.sub(r'@\w+', ' ', text)

    # Menghapus hashtag
    text = re.sub(r'#\w+', ' ', text)

    # Menghapus angka
    text = re.sub(r'\d+', ' ', text)

    # Menghapus emoji dan karakter non-ASCII
    text = text.encode('ascii', 'ignore').decode('ascii')

    # Menghapus tanda baca dan karakter khusus
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # Menghapus spasi berlebih
    text = re.sub(r'\s+', ' ', text).strip()

    return text


# Menerapkan Data Cleaning
df_id['clean_text'] = df_id['full_text'].apply(data_cleaning)

# Menampilkan hasil sebelum dan sesudah cleaning
pd.set_option('display.max_colwidth', 300)

df_id[['full_text', 'clean_text']].head(10)
```

```python
# Simpan hasil Data Cleaning
df_id.to_csv(
    "dataset_skripsi/dataset_preprocessing_cleaning.csv",
    index=False
)

print("Hasil Data Cleaning berhasil disimpan.")
```

# *Case Folding*

```python
# Case Folding
df_id['case_folding'] = df_id['clean_text'].str.lower()

# Menampilkan hasil Case Folding
pd.set_option('display.max_colwidth', 300)

df_id[['clean_text', 'case_folding']].head(10)
```

```python
# Simpan hasil Case Folding
df_id.to_csv(
    "dataset_skripsi/dataset_preprocessing_case_folding.csv",
    index=False
)

print("Hasil Case Folding berhasil disimpan.")
```

```python
import csv

# Membaca kamus normalisasi 2025–2026
with open(
    "KAMUS-NORMALISASI-Machine-Learning/fikkamusnormalisasikata-2025-2026.csv",
    "r",
    encoding="utf-8"
) as file:
    reader = csv.reader(file)
    data_kamus = list(reader)

# Membuat DataFrame
kamus = pd.DataFrame(
    data_kamus[1:],
    columns=["tidak_baku", "kata_baku"]
)

# Membersihkan tanda ; pada kata baku
kamus["kata_baku"] = kamus["kata_baku"].str.rstrip(";").str.strip()
```

```python
print("Jumlah data kamus:", len(kamus))
print(kamus.columns.tolist())

display(kamus.head(10))
```

```python
# Membuat dictionary normalisasi
normalization_dict = (
    kamus.drop_duplicates("tidak_baku")
    .set_index("tidak_baku")["kata_baku"]
    .to_dict()
)
```

```python
# Fungsi normalisasi kata
def normalize_text(text):
    return " ".join(
        normalization_dict.get(word, word)
        for word in text.split()
    )
```

```python
# Menerapkan normalisasi pada data
df_id["normalized_text"] = df_id["case_folding"].apply(normalize_text)
```

```python
# Melihat hasil normalisasi
display(
    df_id[["case_folding", "normalized_text"]].head(10)
)
```

```python
# Simpan dataset hasil text preprocessing
df_id.to_csv(
    "dataset_skripsi/dataset_preprocessing_final.csv",
    index=False
)

print("Dataset hasil text preprocessing berhasil disimpan.")
```
