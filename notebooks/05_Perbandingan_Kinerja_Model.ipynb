```python
import matplotlib.pyplot as plt
import numpy as np

metrik   = ["Accuracy", "Macro Precision", "Macro Recall", "Macro F1-score"]
svm      = [0.8446, 0.5433, 0.5568, 0.5401]
indobert = [0.8872, 0.5998, 0.5641, 0.5796]

x = np.arange(len(metrik))
lebar = 0.35

fig, ax = plt.subplots(figsize=(9, 5))
bar1 = ax.bar(x - lebar/2, svm,      lebar, label="SVM")
bar2 = ax.bar(x + lebar/2, indobert, lebar, label="IndoBERT")

ax.set_ylabel("Nilai")
ax.set_title("Perbandingan Kinerja SVM dan IndoBERT")
ax.set_xticks(x)
ax.set_xticklabels(metrik)
ax.set_ylim(0, 1)
ax.legend()

for bars in [bar1, bar2]:
    for b in bars:
        ax.annotate(f"{b.get_height():.4f}",
                    xy=(b.get_x() + b.get_width()/2, b.get_height()),
                    xytext=(0, 3), textcoords="offset points",
                    ha="center", va="bottom", fontsize=8)

plt.tight_layout()
plt.show()
```

```python
kelas       = ["Negatif", "Netral", "Positif"]
svm_f1      = [0.9145, 0.2273, 0.4785]
indobert_f1 = [0.9437, 0.2553, 0.5399]

x = np.arange(len(kelas)); lebar = 0.35
fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(x - lebar/2, svm_f1,      lebar, label="SVM")
ax.bar(x + lebar/2, indobert_f1, lebar, label="IndoBERT")
ax.set_ylabel("F1-score"); ax.set_title("Perbandingan F1-score per Kelas")
ax.set_xticks(x); ax.set_xticklabels(kelas); ax.set_ylim(0, 1); ax.legend()
plt.tight_layout(); plt.show()
```
